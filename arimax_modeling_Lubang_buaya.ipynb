```python
import pandas as pd
import numpy as np
from statsmodels.tsa.stattools import adfuller, acf, pacf
from statsmodels.tsa.arima.model import ARIMA
from statsmodels.stats.diagnostic import acorr_ljungbox
from statsmodels.stats.stattools import jarque_bera
from scipy.stats import boxcox
from scipy.stats import boxcox_normmax
import matplotlib.pyplot as plt
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
```

```python
df_path = "/content/drive/MyDrive/MBKM/Data/Data Aqi Pollutant Meteorologi Lubang buaya.xlsx"
df = pd.read_excel(df_path)
```

```python
y = df['PM2.5']

X = df[['PM10','SO2','CO','O3','NO2']]
```

```python
split = int(len(df) * 0.7)

y_train = y.iloc[:split]
y_test  = y.iloc[split:]

X_train = X.iloc[:split]
X_test  = X.iloc[split:]
```

```python
def adf_test(series):
    result = adfuller(series)
    print("ADF:", result[0])
    print("p-value:", result[1])

print("\nADF train:")
adf_test(y_train)
```

```python
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plot_acf(y_train, ax=plt.gca(), lags=50)
plt.title("ACF Lubang Buaya")

plt.subplot(1, 2, 2)
plot_pacf(y_train, ax=plt.gca(), lags=50, method='ywm')
plt.title("PACF Lubang Buaya")

plt.tight_layout()
plt.savefig(
    "/content/drive/MyDrive/MBKM/Analisis ARIMAX/Gambar/Lubang Buaya/ACF_PACF_Lubang Buaya.jpeg",
    dpi=600,
    format="jpeg",
    bbox_inches="tight"
)
plt.show()
```

```python
y_diff1 = np.diff(np.diff(y_train))
```

```python
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plot_acf(y_diff1, ax=plt.gca(), lags=50)
plt.title("ACF differencing Lubang Buaya")

plt.subplot(1, 2, 2)
plot_pacf(y_diff1, ax=plt.gca(), lags=50, method='ywm')
plt.title("PACF differencing Lubang Buaya")

plt.tight_layout()
plt.savefig(
    "/content/drive/MyDrive/MBKM/Analisis ARIMAX/Gambar/Lubang Buaya/ACF_PACF_differencing_Lubang Buaya.jpeg",
    dpi=600,
    format="jpeg",
    bbox_inches="tight"
)
plt.show()
```
